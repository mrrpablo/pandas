```python
import matplotlib.pyplot as plt

# Sample point labels
points = ['P1', 'P2', 'P3', 'P4', 'P5', 'P6', 'P7', 'P8', 'P9', 'P10']

# Spectral index data from the table
ndvi_14 = [0.28, 0.312, 0.147, 0.108, -0.056, 0.099, 0.189, 0.292, 0.297, 0.31]
ndvi_24 = [0.132, 0.145, 0.052, 0.041, -0.039, 0.063, 0.104, 0.142, 0.171, 0.164]

ndmi_14 = [0.109, 0.216, 0.044, -0.027, 0.101, 0.0039, 0.111, 0.154, 0.153, 0.169]
ndmi_24 = [0.056, 0.167, 0.026, -0.0136, 0.12, 0.0046, 0.065, 0.089, 0.104, 0.097]

ndwi_14 = [-0.245, -0.267, -0.129, -0.141, 0.096, -0.111, -0.166, -0.255, -0.252, -0.266]
ndwi_24 = [-0.12, -0.117, -0.046, -0.047, 0.063, -0.065, -0.095, -0.123, -0.141, -0.131]

# Create subplots
fig, axs = plt.subplots(3, 1, figsize=(12, 8), sharex=True)

# NDVI Plot
axs[0].plot(points, ndvi_14, marker='o', linestyle='-', color='green', label='NDVI 2014')
axs[0].plot(points, ndvi_24, marker='o', linestyle='--', color='limegreen', label='NDVI 2024')
axs[0].set_ylabel('NDVI Value')
axs[0].set_title('NDVI Comparison (2014 and 2024)')
axs[0].legend(loc='upper right')
axs[0].grid(True)

# NDMI Plot
axs[1].plot(points, ndmi_14, marker='o', linestyle='-', color='navy', label='NDMI 2014')
axs[1].plot(points, ndmi_24, marker='o', linestyle='--', color='blue', label='NDMI 2024')
axs[1].set_ylabel('NDMI Value')
axs[1].set_title('NDMI Comparison (2014 and 2024)')
axs[1].legend(loc='upper right')
axs[1].grid(True)

# NDWI Plot
axs[2].plot(points, ndwi_14, marker='o', linestyle='-', color='maroon', label='NDWI 2014')
axs[2].plot(points, ndwi_24, marker='o', linestyle='--', color='red', label='NDWI 2024')
axs[2].set_ylabel('NDWI Value')
axs[2].set_xlabel('Sampling Points of the Spectral Indices of Lagos (P1–P10)')
axs[2].set_title('NDWI Comparison (2014 and 2024)')
axs[2].legend(loc='upper right')
axs[2].grid(True)

# Main title
fig.suptitle('Spectral Indices of Lagos State, Nigeria (2014 and 2024)', fontsize=14)

# Adjust layout
plt.tight_layout(rect=[0, 0.03, 1, 0.93])
plt.show()

```
